```python
class Human:
    # コンストラクタ
    def __init__(self, name, age):
        self.name = name
        self.age = age

    # 名前と年齢を出力するメソッド
    def printinfo(self):
        print("名前:", self.name)
        print("年齢:", self.age)


# インスタンスを変数に代入する
human = Human("太郎", 20)

# メソッドを実行する
human.printinfo()
```
